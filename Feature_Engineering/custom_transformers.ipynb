# Custom Transformers

## 1. What is a Transformer?

A **transformer** is a scikit-learn object that implements a `.transform()` method. It is used to transform input features (e.g., scaling, encoding, feature creation).

Most transformers also implement `.fit()` because they may need to learn something from the training data.

---

## 2. What is a Custom Transformer?

A **custom transformer** is a transformer that we build ourselves when scikit-learn does not provide the required transformation.

We define our own transformation logic inside `.transform()`.

---

## 3. Types of Custom Transformers

### 1. FunctionTransformer (Stateless)

- Used when the transformation **does not need to learn from the data**.
- `fit()` is effectively a no-op (it simply returns `self`).
- Transformation depends only on the input values.

**Example:**
- `x²`, `x³`
- `log(x)`
- Celsius → Fahrenheit conversion.

### 2. Class-based Custom Transformer (Stateful)

- Used when the transformation **needs to learn something from the training data** during `fit()`.
- The learned information is stored as attributes (usually ending with `_`).

**Example:**
- StandardScaler (learns mean and standard deviation).
- Custom imputer (learns median).
- Custom frequency encoder (learns category frequencies).

---

## 4. Important Notes

- `fit()` should always return `self`.
- `transform()` should return the transformed feature matrix.
- Learned attributes should end with `_` (e.g., `mean_`, `median_`).

---

## 5. Limitation / Clarification

**❌ Incorrect:** Custom transformers are applied on the entire dataset.

**✅ Correct:** A custom transformer receives **whatever input is passed to it**.

- It can transform the entire feature matrix `X`.
- It can also transform **selected columns** when used inside a `ColumnTransformer` or after column selection in a `Pipeline`.
