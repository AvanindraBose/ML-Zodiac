# Remember

Label Encoder is Strictly used for Target Column. It should not be used for Input Features.

We generally don't use LabelEncoder on input features because it converts categories into arbitrary integers, which can introduce a false ordinal relationship that wasn't present in the original data.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
root_path = Path().cwd().parent.parent.parent
data_path = root_path / "Datasets" / "customers.csv"
```

```python
df = pd.read_csv(data_path)
```
